```python
import os

# Use the path where the dataset was downloaded and extracted
downloaded_dataset_path = "/root/.cache/kagglehub/datasets/kshitij192/cars-image-dataset/versions/3"
train_dir = os.path.join(downloaded_dataset_path, "Cars Dataset", "train")

image_paths = []
labels = []

for label in os.listdir(train_dir):
    label_dir = os.path.join(train_dir, label)
    if os.path.isdir(label_dir):
        for image_file in os.listdir(label_dir):
            image_path = os.path.join(label_dir, image_file)
            image_paths.append(image_path)
            labels.append(label)

print(f"Found {len(image_paths)} images in the training directory.")
```

```python
def load_data(root_path, img_size=(224, 224), batch_size=32):
    image_paths = glob.glob(os.path.join(root_path, '*/*.jpeg'))

    class_names = sorted(set(os.path.basename(os.path.dirname(p)) for p in image_paths))
    class_to_index = {name: i for i, name in enumerate(class_names)}

    paths = []
    labels = []
    for path in image_paths:
        label_name = os.path.basename(os.path.dirname(path))
        paths.append(path)
        labels.append(class_to_index[label_name])

    # One-hot encode the labels
    labels_one_hot = tf.one_hot(labels, depth=len(class_names))

    dataset = tf.data.Dataset.from_tensor_slices((paths, labels_one_hot))

    def process(path, label):
        image = tf.io.read_file(path)
        image = tf.image.decode_jpeg(image, channels=3)
        image = tf.image.resize(image, img_size)
        image = tf.cast(image, tf.float32) / 255.0  # Normalize to [0, 1]
        return image, label

    dataset = dataset.map(process)
    dataset = dataset.shuffle(buffer_size=len(paths))  # or 1000
    dataset = dataset.batch(batch_size).prefetch(tf.data.AUTOTUNE)


    return dataset, class_names
```

```python
downloaded_dataset_path = "/root/.cache/kagglehub/datasets/kshitij192/cars-image-dataset/versions/3"
train_path = os.path.join(downloaded_dataset_path, "Cars Dataset", "train")
val_path = os.path.join(downloaded_dataset_path, "Cars Dataset", "test") # Assuming 'test' is used as validation here based on previous usage
test_path = "/root/.cache/kagglehub/datasets/kshitij192/cars-image-dataset/versions/3/Cars Dataset/test" # This path seems correct
```

```python
train_datagen = ImageDataGenerator(
    validation_split=0.2 # Added validation split
)
```

```python
train_generator = train_datagen.flow_from_directory(
    train_path, # Use the directory path instead of the list of paths
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical', # Changed to categorical
    subset='training', # Specify this is the training subset
    shuffle=True
)
```

```python
validation_generator = train_datagen.flow_from_directory(
    train_path, # Use train_path for validation split as well
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical', # Changed to categorical
    subset='validation', # Specify this is the validation subset
    shuffle=False
)
```

```python
checkpoint = ModelCheckpoint(
    filepath='best_model.h5',
    monitor='val_loss',
    save_best_only=True,
    mode='min',
    verbose=1
)
```

```python
# Calculate class weights using labels from train_generator
labels = train_generator.classes

class_weights = class_weight.compute_class_weight(
    class_weight='balanced',
    classes=np.unique(labels),
    y=labels
)
class_weights = dict(enumerate(class_weights))

print(class_weights)
```

```python
from tensorflow.keras.applications import EfficientNetB0
import os

# Define the number of classes

# Define the path to the pre-trained weights (you might need to download these)
# For demonstration purposes, I'll assume the weights are in a 'weights' directory
# in the current working directory. You'll need to adjust this path.
# Alternatively, you can let Keras download the weights automatically by setting weights='imagenet'
# resnet_weights_path = '/path/to/your/resnet50_weights_tf_dim_ordering_tf_kernels_notop.h5'
# For this example, we'll use weights='imagenet' and include_top=False to remove the classification layer

base_model = EfficientNetB0(include_top=False, weights='imagenet', pooling='max', input_shape=(224, 224, 3))

# Create a Sequential model and add the base model and a dense layer
model = Sequential()
model.add(base_model)
model.add(Dense(512, activation='relu'))
model.add(Dense(7, activation='softmax'))
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

history = model.fit(
    train_generator,
    steps_per_epoch=train_generator.samples//train_generator.batch_size,
    validation_data=validation_generator,
    validation_steps=validation_generator.samples//validation_generator.batch_size,
    epochs=15,
    callbacks=[checkpoint]
)
```

```python
def load_data(root_path, img_size=(128, 128), batch_size=32):
    image_paths = glob.glob(os.path.join(root_path, '*/*')) # Changed to match any file extension

    class_names = sorted(set(os.path.basename(os.path.dirname(p)) for p in image_paths))
    class_to_index = {name: i for i, name in enumerate(class_names)}

    paths = []
    labels = []
    for path in image_paths:
        label_name = os.path.basename(os.path.dirname(path))
        paths.append(path)
        labels.append(class_to_index[label_name])

    # Do not one-hot encode the labels for binary classification compatibility with the model
    # labels_one_hot = tf.one_hot(labels, depth=len(class_names))

    dataset = tf.data.Dataset.from_tensor_slices((paths, labels)) # Use integer labels

    def process(path, label):
        image = tf.io.read_file(tf.cast(path, tf.string)) # Explicitly cast path to string again
        image = tf.image.decode_jpeg(image, channels=3)
        image = tf.image.resize(image, img_size)
        image = tf.cast(image, tf.float32) / 255.0  # Normalize to [0, 1]
        return image, label

    dataset = dataset.map(process)
    dataset = dataset.shuffle(buffer_size=len(paths))  # or 1000
    dataset = dataset.batch(batch_size).prefetch(tf.data.AUTOTUNE)


    return dataset, class_names
```

```python
model.save("best_model.h5")
best_model = load_model("best_model.h5")
```

```python
test_ds, _ = load_data(test_path, img_size=(224, 224), batch_size=16)
```

```python
y_true = []
y_pred = []

for images, labels in test_ds: # Iterate over the test dataset
    probs = best_model.predict(images)  # shape (batch_size, 7)
    preds = np.argmax(probs, axis=1) # Get the index of the class with the highest probability

    # Add assertion to check for consistent lengths within the batch
    assert len(labels.numpy()) == len(preds), "Inconsistent number of samples in a batch"

    y_true.extend(labels.numpy())
    y_pred.extend(preds)

print(f"Length of y_true: {len(y_true)}")
print(f"Length of y_pred: {len(y_pred)}")
```

```python
print(classification_report(y_true, y_pred))
```

```python
cm = confusion_matrix(y_true, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Blues')
plt.title("Confusion Matrix on Test Set")
plt.show()

```
